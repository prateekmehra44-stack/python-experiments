```python
#Que 1
with open("name.txt", "w") as f:
    n = int(input("Enter number of names: "))
    for i in range(n):
        name = input()
        f.write(name + "\n")
```

```python
count = 0

with open("name.txt", "r") as f:
    for line in f:
        name = line.strip()
        if name and name[0].lower() in "aeiou":
            count += 1

print("Names starting with vowel:", count)
```

```python
longest = ""

with open("name.txt", "r") as f:
    for line in f:
        name = line.strip()
        if len(name) > len(longest):
            longest = name

print("Longest name:", longest)
```

```python
with open("name.txt", "r") as f:
    for line in f:
        print(line.strip())
```

```python
with open("name.txt", "r") as f:
    for line in f:
        data = line.split()
        if len(data) > 1 and float(data[1]) > 10:
            print(data[0])
```

```python
n = int(input())

for i in range(n):
    try:
        a, b = input().split()
        result = int(a) // int(b)
        print(result)
    except ZeroDivisionError as e:
        print("Error Code:", e)
    except ValueError as e:
        print("Error Code:", e)
```
