```python
import numpy as np

# Create 3x3 array
arr = np.array([[10, 20, 30],
                [40, 50, 60],
                [70, 80, 90]])

print("Array:\n", arr)

# Sum of rows
row_sum = np.sum(arr, axis=1)
print("\nSum of each row:", row_sum)

# Sum of columns
col_sum = np.sum(arr, axis=0)
print("Sum of each column:", col_sum)

# Find 2nd maximum element
flat = arr.flatten()          # Convert to 1D array
flat_sorted = np.sort(flat)   # Sort array
second_max = flat_sorted[-2]  # Second largest

print("\nSecond Maximum Element:", second_max)
```

```python
import numpy as np

n = int(input("Enter size of matrix: "))

print("Enter first matrix:")
A = []
for i in range(n):
    row = list(map(int, input().split()))
    A.append(row)

print("Enter second matrix:")
B = []
for i in range(n):
    row = list(map(int, input().split()))
    B.append(row)

A = np.array(A)
B = np.array(B)

C = np.dot(A, B)

print("Result of multiplication:\n", C)
```

```python
import pandas as pd

# Create DataFrame
data = {'X': [78, 85, 96, 80, 86],
        'Y': [84, 94, 89, 83, 86],
        'Z': [86, 97, 96, 72, 83]}

df = pd.DataFrame(data)

print("Original DataFrame:\n", df)

# Element-wise power: X^Y^Z (stepwise)
result = df['X'] ** df['Y']   # First power
result = result ** df['Z']    # Then raise to Z

print("\nResult (Element-wise Power):\n", result)
```

```python
import pandas as pd
import numpy as np

# Sample DataFrame
data = {
    'name': ['Aman', 'Ravi', 'Sita', 'John'],
    'score': [90, np.nan, 75, np.nan],
    'attempts': [1, 2, np.nan, 3]
}

df = pd.DataFrame(data)

print("Original DataFrame:\n", df)

# Find missing values
print("\nMissing values:\n", df.isnull())

# Replace missing values (example: replace with 0)
df_filled = df.fillna(0)

print("\nDataFrame after replacing missing values:\n", df_filled)
```

```python
import matplotlib.pyplot as plt

# Sample data
x = [1, 2, 3, 4, 5]
y = [10, 20, 25, 30, 40]

# 1. Line Plot
plt.figure()
plt.plot(x, y)
plt.title("Line Plot")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.show()

# 2. Bar Chart
plt.figure()
plt.bar(x, y)
plt.title("Bar Chart")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.show()

# 3. Scatter Plot
plt.figure()
plt.scatter(x, y)
plt.title("Scatter Plot")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.show()

# 4. Pie Chart
labels = ['A', 'B', 'C', 'D', 'E']
plt.figure()
plt.pie(y, labels=labels, autopct='%1.1f%%')
plt.title("Pie Chart")
plt.show()

# 5. Histogram
data = [10, 20, 20, 30, 30, 30, 40, 50]
plt.figure()
plt.hist(data)
plt.title("Histogram")
plt.xlabel("Values")
plt.ylabel("Frequency")
plt.show()
```
