```python
class Student:
    def __init__(self, name, sap, p, c, m):
        self.name = name
        self.sap = sap
        self.p = p
        self.c = c
        self.m = m

    def display(self):
        print(self.name, self.sap, self.p, self.c, self.m)

s1 = Student("prateek", 5678900987, 80, 70, 90)
s2 = Student("amit", 245678978, 85, 75, 95)
s3 = Student("rohit", 3456789876, 60, 65, 70)

s1.display()
s2.display()
s3.display()
```

```python
class Student:
    def __init__(self, name, sap_id, phy, chem, maths):
        self.name = name
        self.sap_id = sap_id
        self.phy = phy
        self.chem = chem
        self.maths = maths

    # Display details
    def display(self):
        print("\nName:", self.name)
        print("SAP ID:", self.sap_id)
        print("Physics:", self.phy)
        print("Chemistry:", self.chem)
        print("Maths:", self.maths)

    # Calculate percentage
    def marks_percentage(self):
        total = self.phy + self.chem + self.maths
        percentage = total / 3
        return percentage

    # Result (Pass/Fail)
    def result(self):
        if self.phy > 40 and self.chem > 40 and self.maths > 40:
            return "Pass"
        else:
            return "Fail"


# Function to find class average
def class_average(students):
    total = 0
    for s in students:
        total += s.marks_percentage()
    return total / len(students)


# Main program
students = []

n = int(input("Enter number of students: "))

for i in range(n):
    print(f"\nEnter details of Student {i+1}:")
    name = input("Enter Name: ")
    sap_id = input("Enter SAP ID: ")
    phy = float(input("Enter Physics marks: "))
    chem = float(input("Enter Chemistry marks: "))
    maths = float(input("Enter Maths marks: "))

    s = Student(name, sap_id, phy, chem, maths)
    students.append(s)

# Display details, percentage and result
print("\n--- Student Details ---")
for s in students:
    s.display()
    print("Percentage:", s.marks_percentage())
    print("Result:", s.result())

# Class average
avg = class_average(students)
print("\nClass Average Percentage:", avg)
```

```python
class Person:
    def show_person(self):
        print("This is Person class")

# Single Inheritance
class Student(Person):
    def show_student(self):
        print("This is Student class")

# Multilevel Inheritance
class Monitor(Student):
    def show_monitor(self):
        print("This is Monitor class")

# Multiple Inheritance
class Sports:
    def show_sports(self):
        print("This is Sports class")

class Result(Student, Sports):
    def show_result(self):
        print("This is Result class")

# Hierarchical Inheritance
class Teacher(Person):
    def show_teacher(self):
        print("This is Teacher class")


# Main Program
print("---- Single Inheritance ----")
s = Student()
s.show_person()
s.show_student()

print("\n---- Multilevel Inheritance ----")
m = Monitor()
m.show_person()
m.show_student()
m.show_monitor()

print("\n---- Multiple Inheritance ----")
r = Result()
r.show_person()
r.show_student()
r.show_sports()
r.show_result()

print("\n---- Hierarchical Inheritance ----")
t = Teacher()
t.show_person()
t.show_teacher()
```

```python
# Parent class
class Parent:
    def show(self):
        print("This is Parent class method")

# Child class
class Child(Parent):
    def show(self):   # Overriding the parent method
        print("This is Child class method")

# Main program
obj = Child()
obj.show()
```

```python
class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    # Overloading + operator
    def __add__(self, other):
        x = self.x + other.x
        y = self.y + other.y
        return Point(x, y)

    def display(self):
        print(f"Point(x={self.x}, y={self.y})")


# Creating objects
P1 = Point(10, 20)
P2 = Point(12, 15)

# Adding objects
P3 = P1 + P2

# Display result
P3.display()
```
